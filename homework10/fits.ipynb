```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
```

```python
files = [
    "/Users/sophiachoi/python_decal_fall25/course_assignments/homework10/M13_0001.fits",
    "/Users/sophiachoi/python_decal_fall25/course_assignments/homework10/M13_0002.fits",
    "/Users/sophiachoi/python_decal_fall25/course_assignments/homework10/M13_0003.fits",
    "/Users/sophiachoi/python_decal_fall25/course_assignments/homework10/M13_0004.fits",
    "/Users/sophiachoi/python_decal_fall25/course_assignments/homework10/M13_0005.fits"
]
data = []
for file in files:
    with fits.open(file) as hdul:
        header_list.append(hdul[0].header)
        data.append(hdul[0].data)
```

```python
fig, axs = plt.subplots(1, 5, figsize=(18, 4))
for i in range(5):
    axs[i].imshow(data[i], origin="lower")
plt.show()
```

```python
fig, axs = plt.subplots(1, 5, figsize=(18, 4))
im = axs[0].imshow(data[0], origin="lower", cmap="magma", vmin=500, vmax=600)
for i in range(5):
    axs[i].imshow(data[i], origin="lower", cmap="magma", vmin=500, vmax=600)
fig.colorbar(im, ax=axs, location='right')
plt.show()
#These are much clearer!
```
