```python
not_these = ["Planet", "Ring System?", "Global Magnetic Field?"]

for col in df.columns:
    if col not in not_these:
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.dtypes
#Objects, floats, and integers
#Floats include decimals, integers are integers, and the objects don't contain numbers so they can't be switched to them
```

```python
def attach_units(col, unit, new_col):
    new_values = []
    for val in df[col]:
        new_values.append(val * unit)    
    df[col] = new_values
    df.rename(columns={col: new_col}, inplace=True)
    return df
```

```python
attach_units('Escape Velocity (km/s)', u.km/u.s, 'Escape Velocity (km/s)')
```

```python
print(df['Escape Velocity (km/s)'])
```

```python
attach_units('Perihelion (10^6 km)', 1e6 * u.km, 'Perihelion (km)')
attach_units('Aphelion (10^6 km)', 1e6 * u.km, 'Aphelion (km)')
```

```python
semi_major = (df['Perihelion (km)'] + df['Aphelion (km)']) / 2
idx = df.columns.get_loc('Aphelion (km)')
df.insert(idx + 1, "Semi-Major Axis (km)", semi_major)
```

```python
def to_AU(col):
    if not isinstance(df[col].iloc[0], u.Quantity):
        df[col] = df[col].values * u.km
    df[col] = df[col].apply(lambda x: x.to(u.au))

to_AU('Diameter (km)')
df.rename(columns={'Diameter (km)': 'Diameter (AU)'}, inplace=True)
```

```python
to_AU('Aphelion (km)')
df.rename(columns={'Aphelion (km)': 'Aphelion (AU)'}, inplace=True)
```

```python
to_AU('Perihelion (km)')
df.rename(columns={'Perihelion (km)': 'Perihelion (AU)'}, inplace=True)
```

```python
to_AU('Semi-Major Axis (km)')
df.rename(columns={'Semi-Major Axis (km)': 'Semi-Major Axis (AU)'}, inplace=True)
```

```python
attach_units('Distance from Sun (10^6 km)', 1e6 * u.km , 'Distance from Sun (km)')
```

```python
to_AU('Distance from Sun (km)')
df.rename(columns={'Distance from Sun (km)': 'Distance from Sun (AU)'}, inplace=True)
```
