```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("astro_image (1).csv")
```

```python
data = np.loadtxt("astro_image (1).csv")

plt.imshow(data, origin="lower", cmap = "plasma", vmin = 100, vmax = 200)
plt.colorbar()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

zoom1 = data[400:600, 1000:1200]
zoom2 = data[1000:1400, 800:1200]

axes[0].imshow(zoom1, origin="lower", vmin = 100, vmax = 200)
axes[0].set_title("Zoom 1")

axes[1].imshow(zoom2, origin="lower", vmin = 100, vmax = 200)
axes[1].set_title("Zoom 2")

plt.show()
```
